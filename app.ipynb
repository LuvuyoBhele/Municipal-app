# Input Options and criteria
what do you wnat to compare what what do you wnat to compare them against

```python
n = int(input("How many options do you want to compare? "))
if n > 0:
    options = []
    while len(options) < n:
        option = input(f"Enter option {len(options)+1}: ")
        options.append(option)

m = int(input("How many criteria do you want to compare? "))
if m > 0:
    criteria = []
    while len(criteria) < m:
        criterion = input(f"Enter criterion {len(criteria)+1}: ")
        criteria.append(criterion)
```

# criteria
Now we must look at the type of criteria we're comparing is it quantitative data or qualitative and develop a scale. We will use a known scale for now

# here we create the jugement matrix
## to create this we need to compare all the pairwise options

```python
pairs = MF.pairs(options)
pair_scores = {}
for criteria in criteria:
    for pair in pairs:
        print(f"Comparing {pair[0]} and {pair[1]} in term of {criteria}:")
        if pair[0] == pair[1]:
            score = 1.0
        else:
            score = float(input(f"based on {criteria} on a scale of (1-9) how important is {pair[0]} to {pair[1]}: "))
        pair_scores[pair] = [criteria, score]
```
